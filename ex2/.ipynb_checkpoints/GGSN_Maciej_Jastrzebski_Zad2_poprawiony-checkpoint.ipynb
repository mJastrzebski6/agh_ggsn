# Klasyfikacja kart

Notebook porządkuje eksperymenty CNN i transfer learning dla zbioru kart.
Układ jest sekwencyjny. Komórki można uruchamiać po kolei.

Plan:
1. wczytanie danych i szybka eksploracja,
2. szybkie testy kilku architektur CNN na części batchy,
3. ponowny trening najlepszego CNN na pełnym zbiorze,
4. porównanie z transfer learning,
5. ewaluacja i wizualizacje.

## 1. Importy

```python
import gc
import json
import math
import os
import random
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import tensorflow as tf

from sklearn.metrics import confusion_matrix, classification_report

from tensorflow.keras import Model, Sequential
from tensorflow.keras.callbacks import CSVLogger, EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.layers import (
    BatchNormalization,
    Conv2D,
    Dense,
    Dropout,
    Flatten,
    GlobalAveragePooling2D,
    Input,
    MaxPooling2D,
    SeparableConv2D
)
from tensorflow.keras.models import load_model
from tensorflow.keras.optimizers import Adam, RMSprop
from tensorflow.keras.preprocessing.image import ImageDataGenerator, img_to_array, load_img
from tensorflow.keras.applications import MobileNetV2, VGG16
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input as mobilenet_v2_preprocess
from tensorflow.keras.applications.vgg16 import preprocess_input as vgg16_preprocess

print("tf", tf.__version__)
print("python", os.sys.version.split()[0])
print("gpu", tf.config.list_physical_devices("GPU"))

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["axes.grid"] = True
plt.rcParams["grid.alpha"] = 0.2
sns.set_theme(style="whitegrid")
```

## 2. Ścieżki i parametry

```python
BASE_DIR = Path.cwd()

DATA_DIR = BASE_DIR / "../data/cards-image-datasetclassification"
TRAIN_DIR = DATA_DIR / "train"
VALID_DIR = DATA_DIR / "valid"
TEST_DIR = DATA_DIR / "test"

RESULTS_DIR = BASE_DIR / "results_clean"
CHECKPOINTS_DIR = RESULTS_DIR / "checkpoints"
PLOTS_DIR = RESULTS_DIR / "plots"
LOGS_DIR = RESULTS_DIR / "logs"

for directory in [RESULTS_DIR, CHECKPOINTS_DIR, PLOTS_DIR, LOGS_DIR]:
    directory.mkdir(parents=True, exist_ok=True)

IMG_HEIGHT = 224
IMG_WIDTH = 224
IMG_CHANNELS = 3
IMG_SIZE = (IMG_HEIGHT, IMG_WIDTH)

BATCH_SIZE = 32
TRIAL_EPOCHS = 10
FULL_EPOCHS = 25
TRIAL_FRACTION = 0.35
BASE_LR = 1e-3

assert TRAIN_DIR.exists(), f"Brak TRAIN_DIR: {TRAIN_DIR}"
assert VALID_DIR.exists(), f"Brak VALID_DIR: {VALID_DIR}"
assert TEST_DIR.exists(), f"Brak TEST_DIR: {TEST_DIR}"

CLASSES = sorted([d.name for d in TRAIN_DIR.iterdir() if d.is_dir()])
NUM_CLASSES = len(CLASSES)

print("data", DATA_DIR)
print("classes", NUM_CLASSES)
print("img", IMG_SIZE)
print("batch", BATCH_SIZE)
print("trial_fraction", TRIAL_FRACTION)
```

## 3. Funkcje pomocnicze

```python
def cleanup_memory():
    gc.collect()
    tf.keras.backend.clear_session()

def count_images_in_dir(directory: Path):
    counts = {}
    for class_dir in sorted([d for d in directory.iterdir() if d.is_dir()]):
        counts[class_dir.name] = len(list(class_dir.glob("*.jpg"))) + len(list(class_dir.glob("*.jpeg"))) + len(list(class_dir.glob("*.png")))
    return counts

train_counts = count_images_in_dir(TRAIN_DIR)
valid_counts = count_images_in_dir(VALID_DIR)
test_counts = count_images_in_dir(TEST_DIR)

CLASS_TO_IDX = {name: idx for idx, name in enumerate(CLASSES)}
IDX_TO_CLASS = {idx: name for name, idx in CLASS_TO_IDX.items()}

with open(RESULTS_DIR / "class_to_idx.json", "w", encoding="utf-8") as f:
    json.dump(CLASS_TO_IDX, f, ensure_ascii=False, indent=2)

def make_datagens(preprocess_fn=None, augment=False):
    train_kwargs = {}
    if preprocess_fn is None:
        train_kwargs["rescale"] = 1.0 / 255.0
    else:
        train_kwargs["preprocessing_function"] = preprocess_fn

    valid_kwargs = {}
    if preprocess_fn is None:
        valid_kwargs["rescale"] = 1.0 / 255.0
    else:
        valid_kwargs["preprocessing_function"] = preprocess_fn

    if augment:
        train_kwargs.update({
            "rotation_range": 18,
            "width_shift_range": 0.12,
            "height_shift_range": 0.12,
            "zoom_range": 0.15,
            "shear_range": 0.1,
            "horizontal_flip": True,
            "fill_mode": "nearest",
        })

    train_datagen = ImageDataGenerator(**train_kwargs)
    valid_datagen = ImageDataGenerator(**valid_kwargs)
    test_datagen = ImageDataGenerator(**valid_kwargs)

    return train_datagen, valid_datagen, test_datagen

def make_generators(preprocess_fn=None, augment=False):
    train_datagen, valid_datagen, test_datagen = make_datagens(preprocess_fn=preprocess_fn, augment=augment)

    train_gen = train_datagen.flow_from_directory(
        TRAIN_DIR,
        target_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        classes=CLASSES,
        class_mode="categorical",
        shuffle=True,
        seed=SEED
    )

    valid_gen = valid_datagen.flow_from_directory(
        VALID_DIR,
        target_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        classes=CLASSES,
        class_mode="categorical",
        shuffle=False
    )

    test_gen = test_datagen.flow_from_directory(
        TEST_DIR,
        target_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        classes=CLASSES,
        class_mode="categorical",
        shuffle=False
    )

    return train_gen, valid_gen, test_gen

def make_callbacks(name, patience_es=5, patience_lr=2):
    return [
        ModelCheckpoint(
            filepath=str(CHECKPOINTS_DIR / f"{name}.keras"),
            monitor="val_accuracy",
            mode="max",
            save_best_only=True,
            verbose=1
        ),
        EarlyStopping(
            monitor="val_loss",
            patience=patience_es,
            restore_best_weights=True,
            verbose=1
        ),
        ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.5,
            patience=patience_lr,
            min_lr=1e-6,
            verbose=1
        ),
        CSVLogger(str(LOGS_DIR / f"{name}.csv"))
    ]

def compile_model(model, optimizer_name="adam", lr=1e-3, label_smoothing=0.0):
    if optimizer_name == "adam":
        optimizer = Adam(learning_rate=lr)
    elif optimizer_name == "rmsprop":
        optimizer = RMSprop(learning_rate=lr)
    else:
        raise ValueError(f"Nieznany optimizer: {optimizer_name}")

    loss = tf.keras.losses.CategoricalCrossentropy(label_smoothing=label_smoothing)
    model.compile(optimizer=optimizer, loss=loss, metrics=["accuracy"])
    return model

def plot_history(history, title):
    hist = pd.DataFrame(history.history)
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    axes[0].plot(hist["accuracy"], label="train")
    axes[0].plot(hist["val_accuracy"], label="valid")
    axes[0].set_title("acc")
    axes[0].set_xlabel("epoch")
    axes[0].legend()

    axes[1].plot(hist["loss"], label="train")
    axes[1].plot(hist["val_loss"], label="valid")
    axes[1].set_title("loss")
    axes[1].set_xlabel("epoch")
    axes[1].legend()

    fig.suptitle(title)
    plt.tight_layout()
    plt.show()

def get_best_epoch(history, key="val_accuracy"):
    values = history.history[key]
    return int(np.argmax(values) + 1), float(np.max(values))

def evaluate_model(model, generator, name):
    loss, acc = model.evaluate(generator, verbose=1)
    print(f"{name} test_acc={acc:.4f} test_loss={loss:.4f}")
    return {"model": name, "test_acc": float(acc), "test_loss": float(loss)}

def get_last_conv_layer_name(model):
    for layer in reversed(model.layers):
        if isinstance(layer, (Conv2D, SeparableConv2D)):
            return layer.name
        if isinstance(layer, tf.keras.Model):
            nested = get_last_conv_layer_name(layer)
            if nested is not None:
                return nested
    return None
```

## 7. Architektury CNN

```python
def build_cnn_small(dropout=0.3):
    model = Sequential([
        Input(shape=(IMG_HEIGHT, IMG_WIDTH, IMG_CHANNELS)),
        Conv2D(32, 3, activation="relu", padding="same"),
        MaxPooling2D(),
        Conv2D(64, 3, activation="relu", padding="same"),
        MaxPooling2D(),
        Conv2D(128, 3, activation="relu", padding="same"),
        MaxPooling2D(),
        Flatten(),
        Dense(256, activation="relu"),
        Dropout(dropout),
        Dense(NUM_CLASSES, activation="softmax"),
    ], name="cnn_small")
    return model

def build_cnn_bn(dropout=0.4):
    model = Sequential([
        Input(shape=(IMG_HEIGHT, IMG_WIDTH, IMG_CHANNELS)),
        Conv2D(32, 3, activation="relu", padding="same"),
        BatchNormalization(),
        Conv2D(32, 3, activation="relu", padding="same"),
        MaxPooling2D(),
        Dropout(0.15),

        Conv2D(64, 3, activation="relu", padding="same"),
        BatchNormalization(),
        Conv2D(64, 3, activation="relu", padding="same"),
        MaxPooling2D(),
        Dropout(0.2),

        Conv2D(128, 3, activation="relu", padding="same"),
        BatchNormalization(),
        Conv2D(128, 3, activation="relu", padding="same"),
        MaxPooling2D(),
        Dropout(0.3),

        Flatten(),
        Dense(256, activation="relu"),
        BatchNormalization(),
        Dropout(dropout),
        Dense(NUM_CLASSES, activation="softmax"),
    ], name="cnn_bn")
    return model

def build_cnn_gap(dropout=0.4):
    model = Sequential([
        Input(shape=(IMG_HEIGHT, IMG_WIDTH, IMG_CHANNELS)),
        Conv2D(32, 3, activation="relu", padding="same"),
        BatchNormalization(),
        MaxPooling2D(),

        Conv2D(64, 3, activation="relu", padding="same"),
        BatchNormalization(),
        MaxPooling2D(),

        Conv2D(128, 3, activation="relu", padding="same"),
        BatchNormalization(),
        MaxPooling2D(),

        Conv2D(256, 3, activation="relu", padding="same"),
        BatchNormalization(),
        GlobalAveragePooling2D(),
        Dropout(dropout),
        Dense(128, activation="relu"),
        Dropout(dropout),
        Dense(NUM_CLASSES, activation="softmax"),
    ], name="cnn_gap")
    return model

def build_cnn_sep(dropout=0.35):
    model = Sequential([
        Input(shape=(IMG_HEIGHT, IMG_WIDTH, IMG_CHANNELS)),
        SeparableConv2D(32, 3, activation="relu", padding="same"),
        BatchNormalization(),
        MaxPooling2D(),

        SeparableConv2D(64, 3, activation="relu", padding="same"),
        BatchNormalization(),
        MaxPooling2D(),

        SeparableConv2D(128, 3, activation="relu", padding="same"),
        BatchNormalization(),
        MaxPooling2D(),

        SeparableConv2D(256, 3, activation="relu", padding="same"),
        BatchNormalization(),
        GlobalAveragePooling2D(),
        Dropout(dropout),
        Dense(192, activation="relu"),
        Dropout(dropout),
        Dense(NUM_CLASSES, activation="softmax"),
    ], name="cnn_sep")
    return model

cnn_builders = {
    "cnn_small": build_cnn_small,
    "cnn_bn": build_cnn_bn,
    "cnn_gap": build_cnn_gap,
    "cnn_sep": build_cnn_sep,
}
```

## 8. Konfiguracje do szybkich testów

```python
cnn_trials = [
    {"name": "cnn_small_a", "builder": "cnn_small", "dropout": 0.30, "optimizer": "adam", "lr": 1e-3, "label_smoothing": 0.00, "augment": False},
    {"name": "cnn_small_b", "builder": "cnn_small", "dropout": 0.45, "optimizer": "adam", "lr": 7e-4, "label_smoothing": 0.05, "augment": True},
    {"name": "cnn_bn_a", "builder": "cnn_bn", "dropout": 0.40, "optimizer": "adam", "lr": 1e-3, "label_smoothing": 0.00, "augment": True},
    {"name": "cnn_bn_b", "builder": "cnn_bn", "dropout": 0.50, "optimizer": "adam", "lr": 6e-4, "label_smoothing": 0.05, "augment": True},
    {"name": "cnn_gap_a", "builder": "cnn_gap", "dropout": 0.35, "optimizer": "adam", "lr": 1e-3, "label_smoothing": 0.00, "augment": True},
    {"name": "cnn_sep_a", "builder": "cnn_sep", "dropout": 0.35, "optimizer": "adam", "lr": 1e-3, "label_smoothing": 0.00, "augment": True},
    {"name": "cnn_sep_b", "builder": "cnn_sep", "dropout": 0.45, "optimizer": "rmsprop", "lr": 7e-4, "label_smoothing": 0.05, "augment": True},
]

pd.DataFrame(cnn_trials)
```

## 9. Szybki screening

W tej sekcji modele są trenowane na części batchy z epoki.
To nie daje końcowego wyniku, ale zwykle wystarcza do odrzucenia słabszych konfiguracji.
Potem najlepszy wariant jest trenowany od nowa na pełnym zbiorze.

```python
trial_results = []
best_trial_history = None
best_trial_name = None
best_trial_score = -1.0

for cfg in cnn_trials:
    cleanup_memory()
    print("\n" + "=" * 70)
    print("trial", cfg["name"])
    print("=" * 70)

    train_gen, valid_gen, _ = make_generators(preprocess_fn=None, augment=cfg["augment"])
    model = cnn_builders[cfg["builder"]](dropout=cfg["dropout"])
    model = compile_model(
        model,
        optimizer_name=cfg["optimizer"],
        lr=cfg["lr"],
        label_smoothing=cfg["label_smoothing"]
    )

    steps_per_epoch = max(1, int(len(train_gen) * TRIAL_FRACTION))
    validation_steps = len(valid_gen)

    history = model.fit(
        train_gen,
        validation_data=valid_gen,
        epochs=TRIAL_EPOCHS,
        steps_per_epoch=steps_per_epoch,
        validation_steps=validation_steps,
        callbacks=make_callbacks(f"trial_{cfg['name']}", patience_es=3, patience_lr=1),
        verbose=1
    )

    best_epoch, best_val_acc = get_best_epoch(history, "val_accuracy")
    best_val_loss = float(np.min(history.history["val_loss"]))
    last_val_acc = float(history.history["val_accuracy"][-1])

    row = {
        **cfg,
        "best_epoch": best_epoch,
        "best_val_acc": best_val_acc,
        "last_val_acc": last_val_acc,
        "best_val_loss": best_val_loss,
        "steps_per_epoch": steps_per_epoch,
    }
    trial_results.append(row)

    if best_val_acc > best_trial_score:
        best_trial_score = best_val_acc
        best_trial_name = cfg["name"]
        best_trial_history = history.history.copy()

    del model, history, train_gen, valid_gen
    cleanup_memory()

trial_df = pd.DataFrame(trial_results).sort_values(["best_val_acc", "best_val_loss"], ascending=[False, True]).reset_index(drop=True)
trial_df.to_csv(RESULTS_DIR / "cnn_trials.csv", index=False)
trial_df
```

## 14. Transfer learning

```python
def build_transfer_model(base_name="MobileNetV2", dropout=0.3, dense_units=256, trainable_tail=0):
    if base_name == "MobileNetV2":
        base_model = MobileNetV2(
            include_top=False,
            weights="imagenet",
            input_shape=(IMG_HEIGHT, IMG_WIDTH, IMG_CHANNELS)
        )
        preprocess_fn = mobilenet_v2_preprocess
    elif base_name == "VGG16":
        base_model = VGG16(
            include_top=False,
            weights="imagenet",
            input_shape=(IMG_HEIGHT, IMG_WIDTH, IMG_CHANNELS)
        )
        preprocess_fn = vgg16_preprocess
    else:
        raise ValueError(base_name)

    if trainable_tail <= 0:
        base_model.trainable = False
    else:
        base_model.trainable = True
        for layer in base_model.layers[:-trainable_tail]:
            layer.trainable = False

    inputs = Input(shape=(IMG_HEIGHT, IMG_WIDTH, IMG_CHANNELS))
    x = base_model(inputs, training=False)
    x = GlobalAveragePooling2D()(x)
    x = Dropout(dropout)(x)
    x = Dense(dense_units, activation="relu")(x)
    x = Dropout(dropout)(x)
    outputs = Dense(NUM_CLASSES, activation="softmax")(x)

    model = Model(inputs, outputs, name=f"{base_name.lower()}_transfer")
    return model, preprocess_fn

transfer_trials = [
    {"name": "mobilenet_head", "base": "MobileNetV2", "dropout": 0.30, "dense_units": 256, "trainable_tail": 0, "lr": 1e-3},
    {"name": "mobilenet_ft", "base": "MobileNetV2", "dropout": 0.30, "dense_units": 256, "trainable_tail": 25, "lr": 3e-4},
    {"name": "vgg16_head", "base": "VGG16", "dropout": 0.40, "dense_units": 256, "trainable_tail": 0, "lr": 8e-4},
]
pd.DataFrame(transfer_trials)
```

## 15. Szybkie testy transfer learning

```python
transfer_results = []

for cfg in transfer_trials:
    cleanup_memory()
    print("\n" + "=" * 70)
    print("transfer", cfg["name"])
    print("=" * 70)

    model, preprocess_fn = build_transfer_model(
        base_name=cfg["base"],
        dropout=cfg["dropout"],
        dense_units=cfg["dense_units"],
        trainable_tail=cfg["trainable_tail"]
    )
    model = compile_model(model, optimizer_name="adam", lr=cfg["lr"], label_smoothing=0.0)

    train_gen_t, valid_gen_t, _ = make_generators(preprocess_fn=preprocess_fn, augment=True)

    history = model.fit(
        train_gen_t,
        validation_data=valid_gen_t,
        epochs=TRIAL_EPOCHS,
        steps_per_epoch=max(1, int(len(train_gen_t) * TRIAL_FRACTION)),
        validation_steps=len(valid_gen_t),
        callbacks=make_callbacks(f"transfer_{cfg['name']}", patience_es=3, patience_lr=1),
        verbose=1
    )

    best_epoch, best_val_acc = get_best_epoch(history, "val_accuracy")
    transfer_results.append({
        **cfg,
        "best_epoch": best_epoch,
        "best_val_acc": best_val_acc,
        "best_val_loss": float(np.min(history.history["val_loss"]))
    })

    del model, history, train_gen_t, valid_gen_t
    cleanup_memory()

transfer_df = pd.DataFrame(transfer_results).sort_values(["best_val_acc", "best_val_loss"], ascending=[False, True]).reset_index(drop=True)
transfer_df.to_csv(RESULTS_DIR / "transfer_trials.csv", index=False)
transfer_df
```

## 21. Grad-CAM

```python
def make_gradcam_heatmap(model, img_array, last_conv_layer_name):
    grad_model = Model(
        inputs=model.input,
        outputs=[model.get_layer(last_conv_layer_name).output, model.output]
    )

    with tf.GradientTape() as tape:
        conv_output, preds = grad_model(img_array, training=False)
        pred_index = tf.argmax(preds[0])
        class_channel = preds[:, pred_index]

    grads = tape.gradient(class_channel, conv_output)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    conv_output = conv_output[0]
    heatmap = tf.reduce_sum(conv_output * pooled_grads, axis=-1)

    heatmap = tf.maximum(heatmap, 0)
    denom = tf.reduce_max(heatmap)
    if float(denom) > 0:
        heatmap = heatmap / denom
    return heatmap.numpy()

def show_gradcam(model, img_path):
    last_conv_name = get_last_conv_layer_name(model)
    assert last_conv_name is not None, "Brak warstwy konwolucyjnej"

    img = load_img(img_path, target_size=IMG_SIZE)
    img_array = img_to_array(img).astype("float32") / 255.0
    img_batch = np.expand_dims(img_array, axis=0)

    preds = model.predict(img_batch, verbose=0)[0]
    pred_idx = int(np.argmax(preds))
    pred_name = IDX_TO_CLASS[pred_idx]
    pred_prob = float(preds[pred_idx])

    heatmap = make_gradcam_heatmap(model, img_batch, last_conv_name)
    heatmap = cv2.resize(heatmap, (IMG_WIDTH, IMG_HEIGHT))
    heatmap = np.uint8(255 * heatmap)
    heatmap_color = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)
    heatmap_color = cv2.cvtColor(heatmap_color, cv2.COLOR_BGR2RGB)

    base_img = np.array(img).astype("uint8")
    overlay = cv2.addWeighted(base_img, 0.65, heatmap_color, 0.35, 0)

    fig, axes = plt.subplots(1, 3, figsize=(12, 4))
    axes[0].imshow(base_img)
    axes[0].set_title("img")
    axes[0].axis("off")

    axes[1].imshow(heatmap_color)
    axes[1].set_title("cam")
    axes[1].axis("off")

    axes[2].imshow(overlay)
    axes[2].set_title(f"{pred_name} {pred_prob:.2f}")
    axes[2].axis("off")

    plt.tight_layout()
    plt.show()

sample_ok_path = random.choice(test_gen.filepaths)
show_gradcam(best_cnn, sample_ok_path)
```

## 22. Filtry pierwszej warstwy

```python
first_conv = None
for layer in best_cnn.layers:
    if isinstance(layer, (Conv2D, SeparableConv2D)):
        first_conv = layer
        break

assert first_conv is not None, "Brak warstwy konwolucyjnej"

weights = first_conv.get_weights()[0]
w_min, w_max = weights.min(), weights.max()
weights = (weights - w_min) / (w_max - w_min + 1e-8)

n_show = min(16, weights.shape[-1])
fig, axes = plt.subplots(4, 4, figsize=(8, 8))
axes = axes.ravel()

for i in range(n_show):
    kernel = weights[:, :, :, i]
    if kernel.shape[-1] == 3:
        axes[i].imshow(kernel)
    else:
        axes[i].imshow(kernel[:, :, 0], cmap="gray")
    axes[i].set_title(f"f{i}", fontsize=8)
    axes[i].axis("off")

for ax in axes[n_show:]:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

## 24. Funkcje użytkowe

```python
def predict_one_image(model, img_path, top_k=5):
    img = load_img(img_path, target_size=IMG_SIZE)
    arr = img_to_array(img).astype("float32") / 255.0
    arr = np.expand_dims(arr, axis=0)

    preds = model.predict(arr, verbose=0)[0]
    top_idx = np.argsort(preds)[-top_k:][::-1]

    rows = [{"class": IDX_TO_CLASS[i], "prob": float(preds[i])} for i in top_idx]
    out = pd.DataFrame(rows)

    plt.figure(figsize=(4, 4))
    plt.imshow(img)
    plt.axis("off")
    plt.title("img")
    plt.show()

    return out

def export_history(history, path):
    pd.DataFrame(history.history).to_csv(path, index=False)
    print("saved", path)

# przykład
# export_history(best_cnn_history, RESULTS_DIR / "best_cnn_history.csv")
# predict_one_image(best_cnn, test_gen.filepaths[0])
```
